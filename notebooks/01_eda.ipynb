```python
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(9, 4))
sns.countplot(data=df, x='Class', order=df['Class'].value_counts().index, palette='crest')
plt.title('Số lượng mẫu của từng loại hạt đậu (Class Distribution)')
plt.xlabel('Loại đậu')
plt.ylabel('Số lượng mẫu')
plt.show()
```

```python
feature_columns = df.select_dtypes(include="number").columns

df[feature_columns].hist(figsize=(15, 12), bins=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Class', y='Area', palette='Set2')
plt.title('Boxplot phát hiện Outlier của Area theo từng loại đậu')
plt.show()
```

```python
plt.figure(figsize=(14, 10))
corr_matrix = df[feature_columns].corr()

sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Ma trận tương quan giữa 16 đặc trưng (Correlation Heatmap)')
plt.show()
```
